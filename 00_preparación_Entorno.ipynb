```python
%sql
CREATE CATALOG IF NOT EXISTS aerolinea;
USE CATALOG aerolinea;
```

## Creación de esquemas arquitectura medallion

```python
%sql
CREATE SCHEMA IF NOT EXISTS aerolinea.fuentes;
CREATE SCHEMA IF NOT EXISTS aerolinea.bronze;
CREATE SCHEMA IF NOT EXISTS aerolinea.silver;
CREATE SCHEMA IF NOT EXISTS aerolinea.gold;
CREATE VOLUME IF NOT EXISTS aerolinea.fuentes.landing
```

## Creación de volumenes 

```python
%sql
CREATE VOLUME IF NOT EXISTS aerolinea.fuentes.landing
```

## Verificación de esquemas y volumen

```python
%sql
SHOW SCHEMAS IN aerolinea;
SHOW VOLUMES IN aerolinea.fuentes;
```
